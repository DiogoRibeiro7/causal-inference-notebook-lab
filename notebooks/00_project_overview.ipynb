## Causal workflow

A good causal analysis should answer these questions:

1. What is the unit of analysis?
2. What is the treatment?
3. What is the outcome?
4. What is the estimand?
5. What assumptions identify the estimand?
6. Which estimator is appropriate?
7. What diagnostics support the design?
8. How sensitive is the conclusion?

## Estimands used in this project

- **ATE:** average treatment effect across all units.
- **ATT:** average treatment effect among treated units.
- **CATE:** conditional average treatment effect for a subgroup or covariate profile.
- **ITE:** individual treatment effect. Usually not directly observable in real data.

The notebooks focus mostly on ATE and CATE.
